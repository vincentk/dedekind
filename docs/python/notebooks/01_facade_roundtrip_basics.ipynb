# Dedekind Python Facade

The facade is handle-only over the native C++ core: Python holds handles plus
accessibility bits, and membership runs the native characteristic morphism χ
across the nanobind boundary. This notebook imports the package, exercises the
facade roundtrip helpers, and exhibits the canonical sets.

```python
import dedekind
```

## Facade roundtrip helpers

Native set / path normalisation across the boundary. The asserts pin the
contract; the values are shown by their `repr`.

```python
assert dedekind.ordered_set_roundtrip([3, 1, 2, 2]) == [1, 2, 3]
assert dedekind.unordered_set_roundtrip([4, 2, 4, 1]) == [1, 2, 4]
assert dedekind.path_from_range([2, 4, 6, 8]) == [2, 4, 6, 8]

(
    dedekind.ordered_set_roundtrip([3, 1, 2, 2]),
    dedekind.unordered_set_roundtrip([4, 2, 4, 1]),
    dedekind.path_from_range([2, 4, 6, 8]),
)
```

## The README exhibit: canonical sets 𝔹 and ℕ, native

The *meat*, not the skin: membership runs the **native** characteristic
morphism χ from the C++ core across the nanobind boundary, and the `repr` is
the mathy blackboard symbol. Composition (`&` / `|` / `~`) is the deferred
next slice (gated on the value-first reducer, #922).

```python
from dedekind import 𝔹, ℕ, Nat, ext  # public re-export; 𝔹→B, ℕ→N via NFKC

# Membership is decided by the native C++ χ across the boundary, not a Python
# reimplementation.  The asserts pin the contract; the money shot is -7 ∉ Nat.
assert True in 𝔹
assert 4 in ℕ and 4 in Nat
assert -7 not in Nat
assert ext([-2, -1, 0, 1, 2]) == {0, 1, 2}

# The canonical sets carry their blackboard repr across the nanobind boundary;
# ext materialises a finite universe through the native χ into a Python set.
𝔹, ℕ, Nat, ext([-2, -1, 0, 1, 2])
```
