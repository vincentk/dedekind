# Dedekind Python Facade: Happy Path
This notebook runs the MVP facade on valid inputs, prints the returned values, and asserts the expected results.

## Payload
Inputs: ordered `[3, 1, 2, 2]`, unordered `[4, 2, 4, 1]`, path `[2, 4, 6, 8]`.
Expected outputs: `[1, 2, 3]`, `[1, 2, 4]`, and `[2, 4, 6, 8]`.

```python
from pathlib import Path

def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "pyproject.toml").exists():
            return candidate
    raise FileNotFoundError("Could not locate repo root containing pyproject.toml")

repo_root = find_repo_root(Path.cwd())

try:
    import dedekind
except ModuleNotFoundError as err:
    raise ModuleNotFoundError(
        "dedekind is not installed in this notebook kernel. "
        f"From {repo_root}, run `python -m pip install -e .` "
        "or use `make integration-test` for the CI-style execution path."
    ) from err

print("happy-path: running facade calls")

ordered = dedekind.ordered_set_roundtrip([3, 1, 2, 2])
unordered = dedekind.unordered_set_roundtrip([4, 2, 4, 1])
path = dedekind.path_from_range([2, 4, 6, 8])

results = {
    "ordered_set_roundtrip": ordered,
    "unordered_set_roundtrip": unordered,
    "path_from_range": path,
}

for label, value in results.items():
    print(f"{label}: {value}")
```

```python
assert ordered == [1, 2, 3]
assert unordered == [1, 2, 4]
assert path == [2, 4, 6, 8]

print("happy-path: assertions passed")
print("notebook-01-ok")
```

## The README exhibit: canonical sets ᵓ9 and ℕ, native

The *meat*, not the skin: membership runs the **native** characteristic
morphism χ from the C++ core across the nanobind boundary, and the `repr` is
the mathy blackboard symbol. Composition (`&` / `|` / `ext`) is the deferred
next slice (gated on the value-first reducer, #922).

```python
from dedekind._dedekind import 𝔹, ℕ  # native canonical sets (source 𝔹→B, ℕ→N via NFKC)

# Membership is decided by the C++ χ, not a Python reimplementation.
assert True in 𝔹
assert 4 in ℕ
assert -7 not in ℕ          # the money shot: χ(-7) = False in C++

print(f"repr: {𝔹!r}  {ℕ!r}")        # mathy blackboard repr across the boundary
print(f"True ∈ 𝔹 : {True in 𝔹}")
print(f"4 ∈ ℕ : {4 in ℕ}   |   -7 ∈ ℕ : {-7 in ℕ}")
print("readme-exhibit-ok")
```
