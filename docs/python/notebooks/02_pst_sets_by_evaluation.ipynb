# Pst by evaluation: sets over the truth chains

The paper's Lwv grammar over 𝔹 and K₃ (`dedekind.pst`). Every handle is a real C++ set; every query is decided in C++ by exhausting the chain. Each cell asserts what it shows, so this notebook is a test.

```python
from dedekind.pst import 𝔸, Ø, η, π, 𝔹, K3, Ternary, Kleene, exists, forall, canonical, lift
F, U, T = Ternary.FALSE, Ternary.UNKNOWN, Ternary.TRUE
assert list(K3) == [F, U, T] and K3.cover(U) == T and K3.cover(T) is None
list(K3), K3.cover(U), K3.cover(T)  # the chain, unfolded by the cover; ⊤ has none
```

## The former, the lattice, and equality decided

```python
S = K3 | (π > F)                      # {x | x > ⊥} = {U, ⊤}
assert canonical(S) == S and repr(S) == '[U, ⊤]'
assert (S == (K3 | (π >= U))) is True   # extensional: the same set, spelled twice
assert ((S | ~S) == K3) is True and ((S & ~S) == Ø(K3)) is True
S, ~S, S & ~S
```

## The #365 showcase, by evaluation

`((x > ⊥) ∪ (x > U)) ∩ (x < U) = Ø` needs distributivity as a rewrite rule on terms; on a finite chain it needs nothing, evaluation distributes.

```python
left = (K3 | (π > F)) | (K3 | (π > U))
right = K3 | (π < U)
assert repr(left) == '[U, ⊤]' and repr(right) == '{⊥}'
assert ((left & right) == Ø(K3)) is True
left, right, left & right
```

## Quantifiers, with `Unknown` as a verdict

A K₃-valued set answers in K₃. `K3.identity` is χ(x) = x; it agrees with itself only to degree U where it is U, and the excluded middle holds to degree U.

```python
H = K3.identity
assert exists(K3, π > F) is True and forall(K3, π > F) is False
assert (H == H) == U
assert (𝔸(Ternary, Kleene) == (H | ~H)) == U       # excluded middle, to degree U
H, H == H, 𝔸(Ternary, Kleene) == (H | ~H)
```

## An L-valued set through its α-cuts

A set stands in a predicate as its characteristic arrow: `H(π)` is χ applied to `π`, a term on the chain, compared by the chain's order.
The α-cut {χ ≥ ℓ} is `K3 | (H(π) >= ℓ)`, the fibre χ⁻¹(ℓ) is `K3 | (H(π) == ℓ)`, and membership is `K3 | (H(π) == T)`: all decidable sets with runs.
A term on its own is not a predicate; it is compared.

```python
cut_U = K3 | (H(π) >= U)
assert (cut_U == (K3 | (π >= U))) is True
assert repr(cut_U) == '[U, ⊤]' and repr(K3 | (H(π) >= T)) == '{⊤}' and repr(K3 | (H(π) == U)) == '{U}'
cut_U, K3 | (H(π) >= T), K3 | (H(π) == U)
```

## Two species, one chain: mixed in their join

A Boolean set beside a K₃-valued one is met and compared in K₃ (𝔹 embeds into K₃ along the dominance).

```python
M = S & H                                   # {U, ⊤} ∧ (χ = id): U at U, ⊤ at ⊤
assert not M.is_decidable and M(U) == U and M(T) == T
assert (S == H) == U and (lift(S) == (S & 𝔸(Ternary, Kleene))) == T
M, S == H
```
