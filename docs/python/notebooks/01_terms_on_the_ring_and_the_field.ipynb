# Terms on the canonical ring and field

The two universes a Python user meets first, each as large as a machine word allows: `u64`, the ring ℤ/2⁶⁴, and `𝔽ₚ`, the prime field with p = 2⁶⁴ − 59 (`dedekind.pst`). A term names no carrier; `A(carrier) | term` binds it and checks it then. Each cell asserts what it shows, so this notebook is a test.

```python
from dedekind.jlt import succ
from dedekind.pst import A, image, u64, 𝔽ₚ, π, π1, π2  # 𝔽ₚ is the name Fp (NFKC)

assert u64(2**64 - 1) + 1 == u64(0)        # the ring wraps
assert 𝔽ₚ(1) / 3 * 3 == 1                   # the field divides
assert u64.is_ring and not u64.is_field and 𝔽ₚ.is_field   # the C++ concepts' verdict
u64(2**64 - 1) + 1, 𝔽ₚ(1) / 3, 𝔽ₚ(-1)
```

## One set, one normal form

`{x > 3} ∩ {x < 5} = {4}`, on pivots that exist only at run time. A set built from comparisons of `π` with constants is read as its run list, so equality, `runs()`, iteration and slicing are the library's.

```python
U = A(u64)
region = (U | (π > 3)) & (U | (π < 5))
assert region == U | (π == 4)
assert region.runs() == [(4, 4)] and list(region) == [4]
assert (~(U | (π == 5))).runs() == [(0, 4), (6, 2**64 - 1)]
region, ~(U | (π == 5))
```

## The laws decide the syntax

A term may use the operators the carrier's certified algebras have: `/` is a field's, `<` an order's. Everything else is refused when the set is built, not when it is read.

```python
def refused(build, error):
    try:
        build()
    except error:
        return True
    return False

assert refused(lambda: A(u64) | (π / 2 == 1), TypeError)   # 2 is not a unit of ℤ/2⁶⁴
assert refused(lambda: A(𝔽ₚ) | (π < 3), TypeError)        # 𝔽ₚ has no order
assert refused(lambda: A(u64) | (π > -1), ValueError)     # the word cannot hold -1
assert refused(lambda: π % 0, ValueError)                 # refused when written

third = A(𝔽ₚ) * A(𝔽ₚ) | (π2 == π1 / 3)                    # a field's division by a constant
assert (3, 1) in third and (1, int(𝔽ₚ(1) / 3)) in third
third
```

## The successor's image wraps

On the word the successor sends 2⁶⁴ − 1 to 0, so the image of a ray gains the point 0: the run list translates.

```python
assert image(succ(int), U | (π > 5)) == U | ((π > 6) | (π == 0))
image(succ(int), U | (π > 5))
```

## The ring's normal form: polynomials

A chain reads a term as run lists; a ring reads it as a polynomial over the carrier, the finite function monomial ↦ coefficient.
Two terms are equal in the free algebra iff their normal forms are.
The coefficients live on the carrier, so on `u64` they wrap, and on `𝔽ₚ` a division by a nonzero constant is a coefficient.
`//`, `%` and a division by a term have no polynomial: they are refused, never rewritten.

```python
assert U.normal_form((π1 + π2) * (π1 + π2)) == U.normal_form(π1 * π1 + 2 * π1 * π2 + π2 * π2)
assert U.normal_form(π * (2**64 - 1) + π) == U.normal_form(0 * π)   # the coefficients wrap
assert A(𝔽ₚ).normal_form(π * 3 / 3) == A(𝔽ₚ).normal_form(π)           # 1/3 is a coefficient
assert refused(lambda: U.normal_form(π % 2), TypeError)
U.normal_form((π1 + π2) * (π1 + π2))
```
